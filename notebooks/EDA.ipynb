# Loan Approval Prediction - Exploratory Data Analysis


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set plot style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

## 1. Data Loading


```python
df = pd.read_csv('../dataset/loan_data.csv')
print(f"Dataset Shape: {df.shape}")
print("-" * 50)
print(df.info())
print("-" * 50)
df.head()
```

## Summary Statistics


```python
df.describe(include='all').T
```

## 2. EDA Visualizations


### Loan Approval Distribution


```python
sns.countplot(x='Loan_Status', data=df, palette='Set2')
plt.title('Loan Approval Distribution')
plt.show()
```

### Applicant Income Distribution


```python
sns.histplot(df['ApplicantIncome'], bins=50, kde=True, color='blue')
plt.title('Applicant Income Distribution')
plt.show()
```

### Co-applicant Income Distribution


```python
sns.histplot(df['CoapplicantIncome'], bins=50, kde=True, color='green')
plt.title('Co-applicant Income Distribution')
plt.show()
```

### Loan Amount Distribution


```python
sns.histplot(df['LoanAmount'].dropna(), bins=50, kde=True, color='orange')
plt.title('Loan Amount Distribution')
plt.show()
```

### Credit History vs Loan Status


```python
sns.countplot(x='Credit_History', hue='Loan_Status', data=df, palette='Set1')
plt.title('Credit History vs Loan Status')
plt.show()
```

### Education vs Loan Status


```python
sns.countplot(x='Education', hue='Loan_Status', data=df, palette='Set2')
plt.title('Education vs Loan Status')
plt.show()
```

### Property Area vs Loan Status


```python
sns.countplot(x='Property_Area', hue='Loan_Status', data=df, palette='viridis')
plt.title('Property Area vs Loan Status')
plt.show()
```

### Box Plots for Numerical Features


```python
num_cols = ['ApplicantIncome', 'CoapplicantIncome', 'LoanAmount']
for col in num_cols:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x='Loan_Status', y=col, data=df, palette='Set3')
    plt.title(f'Box Plot of {col} by Loan Status')
    plt.show()
```

### Pair Plot for Important Variables


```python
sns.pairplot(df[['ApplicantIncome', 'CoapplicantIncome', 'LoanAmount', 'Credit_History', 'Loan_Status']].dropna(), hue='Loan_Status', palette='husl')
plt.show()
```

### Correlation Heatmap


```python
plt.figure(figsize=(10, 8))
# Compute correlation only on numeric columns
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```
